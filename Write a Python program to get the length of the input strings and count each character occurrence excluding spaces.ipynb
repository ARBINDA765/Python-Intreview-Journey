Write a Python program to get the length of the input strings and count each character occurrence excluding spaces:



```python
# Approach Solution-1
input_string="abc abc zx"
from collections import Counter
counter=Counter(input_string)
def string_length_and_char_count(input_string):
    counter=Counter(input_string)
    total_length=[value for key,value in counter.items() if not key==" "]
    char_count=[{key,value}for key,value in counter.items() if not key==" "]
    return sum(total_length),char_count
    
```

```python
input_string="abc abc zx"
string_length_and_char_count(input_string)
```

```python
#Appraoch Solution 2
#without using counter
def string_length_and_char_count_1(input_string):
    char_count = {}
    length = 0
    for char in input_string.replace(" ", ""):
        length += 1
        char_count[char] = char_count.get(char, 0) + 1
    return length, char_count
```

```python
string_length_and_char_count_1(input_string)
```
