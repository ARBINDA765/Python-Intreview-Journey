```python
#Remove characters from the first string which are present in the second string

#Input:
string1 = "computer"
string2 = "cat"
#Output: “ompuer”
```

```python
#Approach-1

common_string=set(string1) & set(string2)
output=""
for i in string1:
    if i not in common_string:
        output+=i
        
output     
```

```python
#Approach-2
common_string=set(string1) & set(string2)
"".join([i for i in string1 if i not in common_string])
```
