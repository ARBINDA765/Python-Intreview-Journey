```python
#Time complexicity on(n)
#space complexicity O(1)
def two_sum(input_list, target):
    i = 0
    j = len(input_list) - 1

    while i <= j:
        if input_list[i] + input_list[j] == target:
            print(input_list[i], input_list[j])
            return True
        elif input_list[i] + input_list[j] < target:
            i += 1
        else:
            j -= 1  # Decrement j instead of incrementing
    return False
```

```python
input_list=[1,4,7,8,9,11]
target=13
two_sum(input_list, target)
```

```python
#Time complexicity on(n^2)
#space complexicity O(1)
def two_sum_1(input_list,target):
    for i in range(len(input_list)):
        for j in range(i+1,len(input_list)):
            if input_list[i]+input_list[j]==target:
                print(input_list[i],input_list[j])
                return True
    return False
    
```
