```python
#flatten a nested list

input_list=[1,2,[3,4,5,6],9,10,11,0,[23,45,78],[100,90]]
```

```python
flat_list=[]
for sublist in input_list:
    if (type(sublist)==int):
        flat_list.append(sublist)
    elif type(sublist)==list:
        for item in sublist:
            flat_list.append(item)
```

```python
input_list_1=[[1,2,3],[10,20,30],[100,200,300,400]]
def flatten_list(input_list_1):
    flat_list_1=[item for sublist in input_list_1 for item in sublist]
    return flat_list_1

flatten_list(input_list_1)
```

```python
from itertools import chain
list_of_lists = [[1, 2, 3], [4, 5, 6], [7], [8, 9]]
flatten_list = list(chain.from_iterable(list_of_lists))
print(flatten_list)  # Output: [1, 2, 3, 4, 5, 6, 7, 8, 9]
```
