```python
#Option-1
def reverse_string(s):
    reverse_string=""
    for char in s:
        reverse_string=char+reverse_string
    return reverse_string

s="aAbc"
reverse_string(s)
        
```

```python
#Option-2
def reverse_string_(s):
    if len(s) == 0:
        return ''
    else:
        return reverse_string_(s[1:]) + s[0]
  
```

```python
s="xyz"
reverse_string_(s)
```
