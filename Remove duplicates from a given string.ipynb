```python
#Appraoch-1

#Native Approach
#Time Complexity: O(n * n) 
#Auxiliary Space: O(1), Keeps the order of elements the same as the input. 

input_str ="HappyNewYear"
ans=""
for i in input_str:
    if i  not in ans:
        ans+=i
print(ans)
```

```python
#Approach-2

#Remove duplicates from a given string using Hashing
def remove_duplicates(s):
    exists = set()
    result = []

    for char in s:
        if char not in exists:
            result.append(char)
            exists.add(char)

    return ''.join(result)


s = "happynewYear"
print(remove_duplicates(s))

```

```python
#Approach-3

# This creates an ordered dictionary where the keys are the unique characters in the order they appear in the input string. 
#Since dictionaries don't allow duplicate keys, this effectively removes duplicates.
s = "happynewYear"
from collections import OrderedDict
def remove_duplicates(s):
    return "".join(OrderedDict.fromkeys(s))
remove_duplicates(s)
```
