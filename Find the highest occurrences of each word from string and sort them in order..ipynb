```python
#Using Sort word
from collections import Counter
def sort_words_by_frequency(text):
    words=text.split()
    word_count=Counter(words)
    sorted_words=sorted(word_count.items(),key=lambda item:item[1],reverse=True)
    #sorted(iterable, key=key, reverse=reverse)
    return sorted_words
text = "apple banana apple orange banana apple orange apple"
sorted_word_frequencies = sort_words_by_frequency(text)
print(sorted_word_frequencies)

    
```

```python
#Withoutusing the counter 
def sort_words_by_frequency(text):
    # Split the text into words
    words = text.split()
    
    # Initialize an empty dictionary to count word occurrences
    word_count = {}
    
    # Count the occurrences of each word
    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1
    
    # Convert the dictionary to a list of tuples and sort by frequency (highest first)
    sorted_words = sorted(word_count.items(), key=lambda item: item[1], reverse=True)
    
    return sorted_words

# Example usage
text = "apple banana apple orange banana apple orange apple"
sorted_word_frequencies = sort_words_by_frequency(text)
print(sorted_word_frequencies)
```

```python
#using get method in dict
def sort_words_by_frequency(text):
    # Split the text into words
    words = text.split()
    
    # Initialize an empty dictionary to count word occurrences
    word_count = {}
    
    # Count the occurrences of each word
    for word in words:
        # Use .get(word, 0) + 1 to count occurrences
        word_count[word] = word_count.get(word, 0) + 1
    
    # Convert the dictionary to a list of tuples and sort by frequency (highest first)
    sorted_words = sorted(word_count.items(), key=lambda item: item[1], reverse=True)
    
    return sorted_words

# Example usage
text = "apple banana apple orange banana apple orange apple"
sorted_word_frequencies = sort_words_by_frequency(text)
print(sorted_word_frequencies)

```
