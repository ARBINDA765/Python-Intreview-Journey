```python
"""Reordering a list, such as [0, 5, 9, 0, 20, 12, 0], so that all zeros are placed at the end while maintaining the order of non-zero elements.
"""
```

```python
#Approach-1
#Time O(n)+O(n) --two for loop
#space-O(n)-Additional list
A = [5, 6, 0, 4, 6, 0, 9, 0, 8]
ans_list=[]
for i in A:
    if i!=0:
        ans_list.append(i)
for j in A:
    if j==0:
        ans_list.append(j)
        
        
ans_list        
```

```python
#Approach-2
#Time O(n)+O(n) --two for loop
#space-O(1)-Constant Space
A = [5, 6, 0, 4, 6, 0, 9, 0, 8]

# Two pointers approach
non_zero_index = 0

# Move all non-zero elements to the front
for i in range(len(A)):
    if A[i] != 0:
        A[non_zero_index] = A[i]#filling non zero index in the same list
        non_zero_index += 1

# Filling  the remaining positions with zeros--len-Nonzeroindex 
for i in range(non_zero_index, len(A)):
    A[i] = 0

print(A)
```

```python
#Approach-3
#Swapping Logic
#Time-O(n)
#Space-O(1)
A = [1, 9, 8, 4, 0, 0, 2, 7, 0, 6, 0, 9]
j=0
for i in range(len(A)):
    if A[i]!=0:
        A[i],A[j]=A[j],A[i]
        j=j+1
print(A)
```
