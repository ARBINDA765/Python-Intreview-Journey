```python
#using set
l = [3,1,6,2,7,5,4,9,16,14]
#take the start variable and end(maximum value of list) variable 
#create a range with the help of start ,end valaue
#using set range - list = will give those are present in A but not in B
#Return all the values as a list

def missingValue(l):
    start = 0
    end = max(l)
    result = set(range(start,end+1))-set(l)
    return list(result)
missingValue(l)
```
