```python
#Solution-1
#Calculate the Intersection of Two Lists without using Built-in Functions

A=[1,2,3,4,5]
B=[4,5,6,7,8,10,100]
ans_list=[]

for i in A:
    for j in B:
        if i==j:
            ans_list.append(i)
            
ans_list
```

```python
#we have nested list
#using lambda and filter
#step-1 - iterate the sublist 2
#step-2 by lambda funcrion iterrate the list1
#step-3 filter will the apply each sublist of list2
#step-4 converting to list
lst1 = [1, 6, 7, 10, 13, 28, 32, 41, 58, 63]
lst2 = [[13, 17, 18, 21, 32], [7, 11, 13, 14, 28], [1, 5, 6, 8, 15, 16]]

def intersection(lst1,lst2):
    lst3=[list(filter(lambda x:x in lst1 ,sublist)) for sublist in lst2]
    return lst3

intersection(lst1,lst2)
```
