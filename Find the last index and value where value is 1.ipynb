```python
#option-1
#using enumerate function

s='10010101001'
max_index=0
for index,value in enumerate(s):
    if value=='1':
        if index >max_index:
            max_index=index
print({max_index:int(s[max_index])})
```

```python
#Option-2
#using rfind 
s = '1001010100100001'
max_index = s.rfind('1')
print({max_index: int(s[max_index])})
```

```python
#Option 3
s = '10101'
max_index = 0

for index in range(len(s)):
    if s[index] == '1':
        max_index = index

print({max_index: int(s[max_index])})
```
