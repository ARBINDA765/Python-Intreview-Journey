```python
tuples_list = [(1, 'apple'), (3, 'banana'), (2, 'cherry')]

# Implementing bubble sort
n = len(tuples_list)
for i in range(n):
    for j in range(0, n-i-1):
        # Swap if the element found is less than the next element (for descending order)
        if tuples_list[j][0] < tuples_list[j+1][0]:
            tuples_list[j], tuples_list[j+1] = tuples_list[j+1], tuples_list[j]

print(tuples_list)

```
