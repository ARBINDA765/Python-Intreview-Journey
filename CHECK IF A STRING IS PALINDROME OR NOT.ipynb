```python
#CHECK IF A STRING IS PALINDROME OR NOT
```

```python
#Solution 1
def is_palindrome(s):
    return s==s[::-1]
s = "abc"
print(is_palindrome(s))  # Output: True
    
```

```python
#solution - 2
#Using Iterative Method
#we will be targeting to compare the first element and last element and increasing by one on both the side 
#suppose 4 elements are there starting from 0. Hence 0 with -1,1 with -2 to be comapred if len of string is 4
#0,1
def is_palindrome_2(s):
    for i in range(len(s)//2):
        if s[i] != s[-(i+1)]:
            return False
    return True
s = "arar"
print(is_palindrome_2(s))  
        
```

```python
#Solution -3
#just replacing the spaces
def is_palindrome_3(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

# Example usage
s = "Data Engineering Python "
print(is_palindrome_3(s))  
```
