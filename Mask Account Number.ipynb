```python
import pandas as pd

# Sample DataFrame
data = {
    'account_number': ['1234567890', '9876543210', '1122334455']
}
df = pd.DataFrame(data)

def mask_account_number(account_number):
    return '*' * (len(account_number)-1)+account_number[-2:]
df["masked_account_number"]=df["account_number"].apply(mask_account_number)


df
```
