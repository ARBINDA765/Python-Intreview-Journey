Write a Python function to return the result string which should not contain digits and duplicate characters:


```python
#Approach-1
#Using Set
def duplicate_remove_no_digit(input_string):
    seen=set()
    result=[]
    for char in input_string:
        if char.isalpha() and char not in seen:
            result.append(char)
            seen.add(char)
    return "".join(result)

```

```python
input_string="Abc902qwAb"
duplicate_remove_no_digit(input_string)
```

```python
#Approach - 2
#using Counter 
def only_alpha_remove_duplicates(input_string):
    filter_string=''.join([char for char in input_string if not char.isdigit()])
    counter=Counter(filter_string)
    final_list=''.join([key for key,value in counter.items() if value>=1])
    return final_list

```

```python
input_string="Abc902qwAb"
only_alpha_remove_duplicates(input_string)
```

```python
#Approach-3
def remove_duplicates(input_string):
    filtered_string=''.join([char for char in input_string if not char.isdigit() ])
    return ''.join(set(filtered_string))
```

```python
remove_duplicates(input_string)
```
