```python
#DELETE THE 2ND OCCURANCE of the list only and maintain the order 
input_list = ["ab", "xy", "st", "ab", "xy", "ab", "oo"]
seen={}
output_list=[]
for item in input_list:
    if item in seen:
        seen[item] += 1
    else:
        seen[item]=1
    if seen[item] !=2:
        output_list.append(item)
        
    
```

```python
print(output_list)
```

```python
#delete the duplicates from the list:
#dict.fromkeys(input_list) creates a dictionary where each element of input_list becomes a key in the dictionary

input_list = ["ab", "xy", "st", "ab", "xy", "ab", "oo"]
output_list = list(dict.fromkeys(input_list))
print(output_list)
```

```python
#using list compression

input_list = ["ab", "xy", "st", "ab", "xy", "ab", "oo"]

# empty set to keep track of seen elements
seen = set()

# Using list comprehension to include only the first occurrence of each element
output_list = [x for x in input_list if not (x in seen or seen.add(x))]

print(output_list)

```

```python
# Use a loop to create a new list with unique elements while maintaining the order
#without using set

input_list = ["ab", "xy", "st", "ab", "xy", "ab", "oo"]
output_list = []
for x in input_list:
    if x not in output_list:
        output_list.append(x)

print(output_list)
```

```python
#without list compression delete the duplicate records
input_list = ["ab", "xy", "st", "ab", "xy", "ab", "oo"]

# Create an empty set to keep track of seen elements
seen = set()
result=[]
for x in input_list:
    if not x in seen:
        seen.add(x)
        result.append(x)
print(result)
```
