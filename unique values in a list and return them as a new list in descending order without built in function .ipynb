```python
#solution 1
#without using the built in function

input_list=[1,1,8,99,100,100,99,77]
def unique_value_list(input_list):
    ans_list=[]
    for i in input_list:
        if input_list.count(i)>=1 and i not in ans_list:
            ans_list.append(i)
    return ans_list

print("Unique value from the list  {} ".format( input_list),"are",unique_value_list(input_list))      
```

```python
#solution 2
# using the counter function  function

input_list=[1,1,8,99,100,100,99,77]
from collections import Counter
def unique_value_list_1(input_list):
    counting=Counter(input_list)
    final_list=[key for key,value in counting.items() if value >=1]
    return final_list
unique_value_list_1(input_list)
```

```python
#using set
#solution-3

input_list=[1,1,8,99,100,100,99,77]
def unique_value_list_2(input_list):
    return list(set(input_list))
unique_value_list_2(input_list)
```

```python
#using dict find the unique values and sorted in descending order 
#solution 1
input_list=[1,1,8,99,100,100,99,77]
def unique_list_sort(input_list):
    unique_dict={}
    for item in input_list:
        unique_dict[item]=unique_dict.get(item,0)
        unique_dict[item]=unique_dict[item]+1
    return sorted(list(unique_dict.keys()),reverse=True)
   
    
unique_list_sort(input_list)  
```

```python
#using dict find the unique values and sorted in descending order without using build in function
#solution 2

input_list=[1,1,8,99,100,100,99,77]
def unique_list_sort_custom(input_list):
    unique_dict={}
    for item in input_list:
        unique_dict[item]=unique_dict.get(item,0)
        unique_dict[item]=unique_dict[item]+1
    unique_list=list(unique_dict.keys())
    for i in range(0,len(unique_list)):
        for j in range(i+1,len(unique_list)):
            if unique_list[i]<unique_list[j]:
                unique_list[i],unique_list[j]=unique_list[j],unique_list[i]
    return unique_list
                
unique_list_sort_custom(input_list)
```
